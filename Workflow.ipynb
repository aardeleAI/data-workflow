> ⚠️ **Warning — Data Leakage with Cross-Validation**
>
> In this exercise, we scaled the data **before** cross-validation. This is actually a subtle but important mistake: it introduces **data leakage**.
>
> **Why?** When you fit the `StandardScaler` on the full dataset first, the scaling parameters (mean, standard deviation) are computed using information from the validation folds. During cross-validation, those folds are supposed to act as unseen data — but they have already "leaked" into the preprocessing step. This leads to **overly optimistic performance estimates** that won't hold up on truly new data.
>
> **The right approach** is to wrap the scaler and the model together in a `Pipeline`, and *then* pass it to the cross-validation function. This way, scikit-learn fits the scaler **only on the training fold** at each iteration, and merely applies it to the validation fold — which is exactly what happens in a real deployment scenario.
